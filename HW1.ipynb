1 (10pts.) Consider feature vectors, x1 = [2 3 4 8 9]^T and x2 = [2 −3 −4 89]^T

i.) How many features are there in x1 and x2?

In x1 there are 5 features and in x2 there are 4 features

ii.) What are the L1,L2, and L∞ norms for each feature vector?

```python
x1 = (2, 3, 4, 8, 9)
x2 = (2, -3, -4, 89)
def findnorm(x, L):
    if L == 10000:
        return max(abs(feature) for feature in x)
    norm = 0
    for feature in x:
        n = abs(feature)**L
        norm += n
    return norm**(1/L)
print(findnorm(x1, 1))
print(findnorm(x1,2))
print(findnorm(x1,10000))

print(findnorm(x2, 1))
print(findnorm(x2,2))
print(findnorm(x2,10000))
```

3 (10pts.) Calculate 1-norm, 2-norm, ∞-norm, and Frobenius norm of: W = [[1, -1], [2, 0]]

```python
import numpy as np
W = np.array([[1,-1],
              [2, 0]])
def matrixnorm(W, L):
    if L == 1:
        sum_max = 0
        for j in range(W.shape[1]):
            column_sum = 0
            for i in range(W.shape[0]):
                column_sum += abs(W[i,j])
            if column_sum > sum_max:
                sum_max = column_sum
        return sum_max

    if L == "inf":
        max_sum = 0
        for row in W:
            row_sum = 0
            for value in row:
                row_sum += abs(value)
            if row_sum > max_sum:
                max_sum = row_sum
        return max_sum
    if L == 2:
        B = W.T @ W
        eigenvals = np.linalg.eigvalsh(B)
        spect_rad = max(eigenvals)
        return np.sqrt(spect_rad)
    if L == "F":
        sum = 0
        for row in W:
            for entry in row:
                sum += entry**2
        return np.sqrt(sum)
print(matrixnorm(W, 1))
print(matrixnorm(W, 2))
print(matrixnorm(W, "inf"))
print(matrixnorm(W, "F"))
            
        
            
```

5.) (10pts.) In class, we used a neural network to learn how to calculate square root. Use
that code and ideas to teach a neural network how to calculate the 7-th root of any integer
between 1 and 100.

```python
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import numpy as np


#Load dataset
#split into input (X) and output (y)
X = np.arange(1,101)
X = X*1.0
y = X**(1/7)
y = y*1.0

#define keras model
model = Sequential()

model.add(Dense(6,input_dim=1,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.001)
mse = tf.keras.losses.MeanSquaredError(
    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss=mse, optimizer=opt)



#fit the keras model on the dataset (CPU)
model.fit(X,y,epochs=2000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model
predictions = model.predict(X)

#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' % (X[i].tolist(), predictions[i][0], y[i]))
```

```python
import tensorflow as tf
from numpy import loadtxt
from keras.models import Sequential
from keras.layers import Dense
from keras import optimizers
from tensorflow.python.keras.optimizers import *
import numpy as np


#Load dataset
#split into input (X) and output (y)

from sklearn import datasets
iris = datasets.load_iris()
#data array = iris.data
#print(new_iris)
X = iris.data[:, 0]
y = iris.data[:, 2]
X = X*1.0
y = y*1.0

X = X.reshape(-1,1)

#define keras model
model = Sequential()

model.add(Dense(6,input_dim=1,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(6,activation='relu'))
model.add(Dense(1))

#compile the keras model
opt = optimizers.Adam(learning_rate=0.001)
mse = tf.keras.losses.MeanSquaredError(
    reduction=tf.keras.losses.Reduction.SUM)
model.compile(loss=mse, optimizer=opt)



#fit the keras model on the dataset (CPU)
model.fit(X,y,epochs=2000,batch_size=10, verbose=0)
model.summary()

#make class predictions with the model
predictions = model.predict(X)

#summarize the first 10 cases
for i in range(10):
    print('%s => %.2f (expected %.2f)' % (X[i].tolist(), predictions[i][0], y[i]))
```
